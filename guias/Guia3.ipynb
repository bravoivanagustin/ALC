```python
import numpy as np
```

```python
def sistema_sup(U, b):
    M = np.copy(U).astype(float)
    r = np.copy(b).astype(float)

    for i in range(len(M)-1,-1,-1):
        r[i] = r[i]/M[i,i]
        M[i] = M[i]/M[i,i]
        print(r[i],M[i])
        for j in range(i-1,-1,-1):
            r[j] = r[j] - (r[i]*M[j,i])
            M[j] = M[j] - (M[i]*M[j,i])
    return r
```

```python
U = np.array([[1,1,0],[0,2,1],[0,0,1]])
b = np.array([1,2,1])
U@sistema_sup(U,b)
```

```python
def sistema_inf(L, b):
    M = np.copy(L).astype(float)
    r = np.copy(b).astype(float)

    for i in range(0,len(M)):
        r[i] = r[i]/M[i,i]
        M[i] = M[i]/M[i,i]
        print(r[i],M[i])
        for j in range(i+1,len(M)):
            r[j] = r[j] - (r[i]*M[j,i])
            M[j] = M[j] - (M[i]*M[j,i])
    return r
```

```python
L = np.array([[1,0,0],[1,1,0],[2,4,1]])
b = np.array([1,2,1])
L@sistema_inf(L,b)
```
