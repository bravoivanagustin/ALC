```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def borracho(n, p):
    M = np.eye(n)

    for i in range(1,n-1):
        M[i-1][i] = 1-p
        M[i][i] = 0
        M[i+1][i] = p

    return M

M = borracho(20,0.5)
```

```python
def simulacion(M, v):
    for i in range(100000):
        v = M@v
    return v
```

```python
v = [0] + [1/18 for i in range(1,19)] + [0]
v = np.array(v)
```

```python
v = simulacion(M,v)

print(v)
```

```python
e = np.zeros(20)
e[1] = 1
e
```

```python
e = simulacion(M,e)

print(e)
```

```python
M_1 = borracho(20, 0.8)
```

```python
print(simulacion(M_1, v))
```

```python
print(simulacion(M_1, e))
```
