```python
def norma_aprox(A, k):
    x = np.random.rand(k,3)
    s_k = 0
    y = [0]

    for i in range(k):
        n = np.linalg.norm(A@x[i])/np.linalg.norm(x[i])
        s_k = max([s_k, n])
        y.append(s_k)

    x = np.array([i for i in range(k+1)])
    y = np.array(y)
    plt.plot(x,y)

    return (s_k, np.linalg.norm(A))
```

```python
A = np.array([[1,2,3],[6,4,3],[1,5,1]])
norma_aprox(A,1000)
```
