```python
def traza(A):
    res = 0
    for i in range(len(A)):
        res += A[i][i]
    return res
```

```python
def sumatoria(A):
    res = 0
    for i in range(len(A)):
        for j in range(len(A[i])):
            res += A[i][j]
    return res 
```

```python
def positivo_negativo(A):
    pos = 0
    neg = 0
    for i in range(len(A)):
        for j in range(len(A[i])):
            a = A[i][j]
            if a > 0:
                pos += a
            else:
                neg -= a
    if pos > neg:
        return 'Positivos', pos
    else:
        return 'Negativos', neg
```

```python
A = np.array([[0,2,-3,1],[1,0,1,1],[2,0,1,1]])
print(traza(A))
print(sumatoria(A))
print(positivo_negativo(A))
```
